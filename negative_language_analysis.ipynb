```python
from docx import Document
```

```python
doc = Document("tv4.docx")

for i, p in enumerate(doc.paragraphs[:50]):
    print(i, repr(p.text))
```

```python
doc2 = Document("svt.docx")
for i, p in enumerate(doc2.paragraphs[:50]):
    print(i, repr(p.text))
```

```python
import re
import pandas as pd
```

```python
rows = []
paragraphs = [p.text.strip() for p in doc.paragraphs]
for i, text in enumerate(paragraphs):
    match = re.match(r'^\[([0-9:]+)\]\s+(.+)$', text)

    if match and i + 1 < len(paragraphs):
        timestamp = match.group(1)
        speaker = match.group(2)
        speech = paragraphs[i + 1]

        rows.append({
            "timestamp": timestamp,
            "speaker": speaker,
            "text": speech
        })

tv4_df = pd.DataFrame(rows)

tv4_df.head(20)
```

```python
rows2 = []
paragraphs = [p.text.strip () for p in doc2.paragraphs]
for i, text in enumerate(paragraphs):
    match =re.match (r'^\[([0-9:]+)\]\s+(.+)$', text)
    if match and i + 1 < len(paragraphs):
        timestamp = match.group(1)
        speaker = match.group(2)
        speech = paragraphs[i + 1]
        rows2.append({
                "timestamp": timestamp,
                "speaker": speaker,
                "text": speech
            })
svt_df = pd.DataFrame(rows2)

svt_df.head(20)
```

```python
leaders = [
    "Magdalena Andersson  (S)",
    "Ebba Busch  (KD)",
    "Simona Mohamsson  (L)",
    "Ulf Kristersson  (M)",
    "Amanda Lind  (MP)",
    "Jimmie Åkesson  (SD)",
    "Nooshi Dadgostar  (V)",
    "Elisabeth Thand Ringqvist  (C)"
]
tv4_leaders = tv4_df[tv4_df["speaker"].isin(leaders)].copy()
svt_leaders = svt_df[svt_df["speaker"].isin(leaders)].copy()
```

```python
tv4_leaders["debate"] = "TV4"
svt_leaders["debate"] = "SVT"
```

```python
all_debates = pd.concat(
    [tv4_leaders, svt_leaders],
    ignore_index=True
)
```

```python
def tokenize(text):
    return re.findall(r"[a-zåäö]+", text.lower())
all_debates["tokens"] = all_debates["text"].apply(tokenize)

all_debates["word_count_clean"] = all_debates["tokens"].str.len()
```

```python
word_summary_clean = (
    all_debates
    .groupby(["debate", "speaker"])["word_count_clean"]
    .sum()
    .reset_index()
)

word_summary_clean
```

```python
from collections import Counter

all_tokens = []

for tokens in all_debates["tokens"]:
    all_tokens.extend(tokens)

word_freq = Counter(all_tokens)

word_freq.most_common(50)
```

```python
candidate_stems = [
    "problem",
    "kris",
    "misslyck",
    "orol",
    "farl",
    "dålig",
    "sämre",
    "värre",
    "hot",
    "katastrof",
    "kaos",
    "skandal",
    "allvarlig",
    "brist",
    "fel"
]

for stem in candidate_stems:

    matches = []

    for word, count in word_freq.items():
        if word.startswith(stem):
            matches.append((word, count))

    matches = sorted(matches, key=lambda x: x[1], reverse=True)

    print(stem, matches)
```

```python
negative_words = {
    # Problem
    "problem",
    "problemet",
    "problemen",

    # Kris
    "kris",
    "krisen",

    # Misslyckande
    "misslyckande",
    "misslyckats",
    "misslyckat",

    # Oro
    "orolig",
    "oroliga",

    # Fara
    "farliga",
    "farligaste",
    "farligt",

    # Dåligt / sämre
    "dåligt",
    "sämre",

    # Hot
    "hot",
    "hotet",

    # Katastrof
    "katastrof",
    "katastrofala",

    # Kaos
    "kaos",

    # Allvarligt
    "allvarlig",
    "allvarligt",
    "allvarligaste",
    "allvarligare",

    # Brist
    "brist",
    "bristen",
    "brister",

    # Fel
    "fel"
}
```

```python
review_rows = []

for i, row in all_debates.iterrows():

    matched_words = []

    for word in row["tokens"]:
        if word in negative_words:
            matched_words.append(word)

    if matched_words:
        review_rows.append({
            "debate": row["debate"],
            "speaker": row["speaker"],
            "matched_words": matched_words,
            "text": row["text"]
        })

review_df = pd.DataFrame(review_rows)

review_df.head(20)
```

```python
negators = {"inte", "ingen", "inget", "inga", "aldrig"}

hit_rows = []

for i, row in all_debates.iterrows():

    tokens = row["tokens"]

    for j, word in enumerate(tokens):

        if word in negative_words:

            previous_words = tokens[max(0, j-3):j]

            negated = any(
                w in negators
                for w in previous_words
            )

            hit_rows.append({
                "debate": row["debate"],
                "speaker": row["speaker"],
                "word": word,
                "previous_words": previous_words,
                "negated_by_rule": negated,
                "text": row["text"]
            })

hit_review_df = pd.DataFrame(hit_rows)
```

```python
valid_hits = hit_review_df[
    hit_review_df["valid_hit"] == True
]

negative_summary = (
    valid_hits
    .groupby(["debate", "speaker"])
    .size()
    .reset_index(name="negative_hits")
)

negative_summary
```

```python
score_df = negative_summary.merge(
    word_summary_clean,
    on=["debate", "speaker"],
    how="left"
)
```

```python
score_df["negative_per_1000"] = (
    score_df["negative_hits"] /
    score_df["word_count_clean"] *
    1000
)
```

```python
score_df.sort_values(
    ["debate", "negative_per_1000"],
    ascending=[True, False]
)
```

```python
overall_hits = (
    valid_hits
    .groupby("speaker")
    .size()
    .reset_index(name="negative_hits")
)

overall_words = (
    all_debates
    .groupby("speaker")["word_count_clean"]
    .sum()
    .reset_index()
)

overall_score = overall_hits.merge(
    overall_words,
    on="speaker",
    how="left"
)

overall_score["negative_per_1000"] = (
    overall_score["negative_hits"]
    / overall_score["word_count_clean"]
    * 1000
)

overall_score.sort_values(
    "negative_per_1000",
    ascending=False
)
```

```python
def calculate_score(word_set):

    rows = []

    for i, row in all_debates.iterrows():

        tokens = row["tokens"]

        for j, word in enumerate(tokens):

            if word in word_set:

                previous_words = tokens[max(0, j-3):j]

                negated = any(
                    w in negators
                    for w in previous_words
                )

                if not negated:
                    rows.append({
                        "speaker": row["speaker"],
                        "word": word
                    })

    hits = pd.DataFrame(rows)

    hit_summary = (
        hits
        .groupby("speaker")
        .size()
        .reset_index(name="negative_hits")
    )

    result = hit_summary.merge(
        overall_words,
        on="speaker",
        how="left"
    )

    result["negative_per_1000"] = (
        result["negative_hits"]
        / result["word_count_clean"]
        * 1000
    )

    return result.sort_values(
        "negative_per_1000",
        ascending=False
    )
```

```python
version1_words = negative_words.copy()

version2_words = negative_words - {
    "brist",
    "bristen",
    "brister",
    "allvarlig",
    "allvarligt",
    "allvarligare",
    "allvarligaste"
}

version3_words = version2_words - {
    "fel"
}
```

```python
calculate_score(version1_words)
```

```python
rank_df = calculate_score(version1_words).reset_index(drop=True).copy()

rank_df["speaker"] = "████████████"

rank_df["negative_per_1000"] = rank_df["negative_per_1000"].round(2)

rank_df.index = range(1, len(rank_df) + 1)

rank_df
```
